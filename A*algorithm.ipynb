```python
import heapq


def a_star(graph, start, goal, heuristic):


    # Priority queue: (f_score, node)
    open_list = []

    # Put the start node into the priority queue
    heapq.heappush(open_list, (0, start))

    # Cost from start to each node
    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0

    # Estimated total cost
    f_score = {node: float('inf') for node in graph}
    f_score[start] = heuristic(start, goal)

    # Used to reconstruct the final path
    came_from = {}

    while open_list:

        # Get node with lowest f(n)
        current_f, current = heapq.heappop(open_list)

        # If destination is reached
        if current == goal:
            path = []

            while current in came_from:
                path.append(current)
                current = came_from[current]

            path.append(start)
            path.reverse()

            return path, g_score[goal]

        # Explore all neighbors
        for neighbor, cost in graph[current].items():

            # Cost from start to neighbor through current
            tentative_g = g_score[current] + cost

            # If this is a better path
            if tentative_g < g_score[neighbor]:

                came_from[neighbor] = current

                g_score[neighbor] = tentative_g

                f_score[neighbor] = (
                    g_score[neighbor] +
                    heuristic(neighbor, goal)
                )

                heapq.heappush(
                    open_list,
                    (f_score[neighbor], neighbor)
                )

    # No path found
    return None, float('inf')


# ---------------------------------------------------
# HEURISTIC FUNCTION
# ---------------------------------------------------

def heuristic(node, goal):
    """
    Simple heuristic.

    Here we use predefined estimated distances
    to the goal G.
    """

    h = {
        'A': 7,
        'B': 6,
        'C': 2,
        'D': 1,
        'E': 1,
        'F': 0,
        'G': 0
    }

    return h.get(node, 0)


# ---------------------------------------------------
# GRAPH
# ---------------------------------------------------

graph = {
    'A': {
        'B': 2,
        'C': 4
    },

    'B': {
        'A': 2,
        'C': 1,
        'D': 4
    },

    'C': {
        'A': 4,
        'B': 1,
        'D': 2,
        'E': 5
    },

    'D': {
        'B': 4,
        'C': 2,
        'E': 1,
        'F': 5
    },

    'E': {
        'C': 5,
        'D': 1,
        'F': 2
    },

    'F': {
        'D': 5,
        'E': 2,
        'G': 1
    },

    'G': {
        'F': 1
    }
}


# ---------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------

start_node = 'A'
goal_node = 'G'

path, cost = a_star(
    graph,
    start_node,
    goal_node,
    heuristic
)


# ---------------------------------------------------
# OUTPUT
# ---------------------------------------------------

if path:
    print("A* Algorithm")
    print("-------------------------")

    print("Start Node :", start_node)
    print("Goal Node  :", goal_node)

    print("Shortest Path:")
    print(" -> ".join(path))

    print("Total Cost :", cost)

else:
    print("No path found!")
```

```python
# ==========================================
# WEATHER CLASSIFICATION USING LOGISTIC REGRESSION
# ==========================================

# 1. Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# ==========================================
# 2. Load dataset
# ==========================================

df = pd.read_csv("weather.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

# Show first rows
display(df.head())

# ==========================================
# 3. Check columns
# ==========================================

print("\nColumns:")
print(df.columns.tolist())

# ==========================================
# 4. Check target distribution
# ==========================================

print("\nTarget distribution:")
print(df["normalized_label"].value_counts())

# ==========================================
# 5. Remove missing values
# ==========================================

df = df.dropna()

print("\nDataset after removing missing values:")
print(df.shape)

# ==========================================
# 6. Separate features and target
# ==========================================

X = df.drop("normalized_label", axis=1)
y = df["normalized_label"]

print("\nFeatures:")
print(X.columns.tolist())

# ==========================================
# 7. Train/Test Split
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# ==========================================
# 8. Feature Scaling
# ==========================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ==========================================
# 9. Train Logistic Regression
# ==========================================

model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)

print("\nModel training completed!")

# ==========================================
# 10. Prediction
# ==========================================

y_pred = model.predict(X_test_scaled)

# ==========================================
# 11. Accuracy
# ==========================================

accuracy = accuracy_score(y_test, y_pred)

print("\n================================")
print("MODEL PERFORMANCE")
print("================================")

print("Accuracy:", round(accuracy * 100, 2), "%")

# ==========================================
# 12. Classification Report
# ==========================================

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# ==========================================
# 13. Confusion Matrix
# ==========================================

cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm
)

disp.plot()

plt.title("Weather Classification - Confusion Matrix")
plt.show()

# ==========================================
# 14. Save Model and Scaler
# ==========================================

joblib.dump(model, "logistic_model.pkl")
joblib.dump(scaler, "scaler.pkl")

print("\nModel saved as logistic_model.pkl")
print("Scaler saved as scaler.pkl")

# ==========================================
# 15. Example Real-Time Prediction
# ==========================================

# IMPORTANT:
# The values below MUST follow the SAME
# column order as X.columns.

print("\nFeature order:")
print(X.columns.tolist())

# Example: use the first test sample
sample = X_test.iloc[[0]]

sample_scaled = scaler.transform(sample)

prediction = model.predict(sample_scaled)[0]

probabilities = model.predict_proba(sample_scaled)[0]

confidence = max(probabilities) * 100

print("\n================================")
print("REAL-TIME PREDICTION")
print("================================")

print("Predicted class:", prediction)
print("Confidence:", round(confidence, 2), "%")

print("\nClass probabilities:")

for class_value, probability in zip(
    model.classes_,
    probabilities
):
    print(
        "Class",
        class_value,
        ":",
        round(probability * 100, 2),
        "%"
    )
```

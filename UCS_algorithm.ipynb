```python
 import heapq


def uniform_cost_search(graph, start, goal):
    # Priority queue: (cost, node)
    priority_queue = []

    # Add starting node with cost 0
    heapq.heappush(priority_queue, (0, start))

    # Store the minimum cost to reach each node
    cost_so_far = {
        start: 0
    }

    # Store parent nodes to reconstruct the path
    parent = {
        start: None
    }

    while priority_queue:

        # Get node with the lowest path cost
        current_cost, current_node = heapq.heappop(priority_queue)

        # Ignore outdated entries
        if current_cost > cost_so_far[current_node]:
            continue

        # Goal reached
        if current_node == goal:

            # Reconstruct path
            path = []
            node = goal

            while node is not None:
                path.append(node)
                node = parent[node]

            path.reverse()

            return path, current_cost

        # Explore neighboring nodes
        for neighbor, edge_cost in graph[current_node].items():

            new_cost = current_cost + edge_cost

            # If this is a cheaper path
            if neighbor not in cost_so_far or new_cost < cost_so_far[neighbor]:

                cost_so_far[neighbor] = new_cost
                parent[neighbor] = current_node

                heapq.heappush(
                    priority_queue,
                    (new_cost, neighbor)
                )

    # No path exists
    return None, float('inf')


# ----------------------------------------
# GRAPH
# ----------------------------------------

graph = {
    'A': {
        'B': 2,
        'C': 4
    },

    'B': {
        'A': 2,
        'C': 1,
        'D': 4
    },

    'C': {
        'A': 4,
        'B': 1,
        'D': 2,
        'E': 5
    },

    'D': {
        'B': 4,
        'C': 2,
        'E': 1,
        'F': 5
    },

    'E': {
        'C': 5,
        'D': 1,
        'F': 2
    },

    'F': {
        'D': 5,
        'E': 2,
        'G': 1
    },

    'G': {
        'F': 1
    }
}


# ----------------------------------------
# MAIN PROGRAM
# ----------------------------------------

start = 'A'
goal = 'G'

path, cost = uniform_cost_search(
    graph,
    start,
    goal
)


# ----------------------------------------
# OUTPUT
# ----------------------------------------

print("Uniform Cost Search")
print("------------------------")

if path:
    print("Start Node :", start)
    print("Goal Node  :", goal)
    print("Path       :", " -> ".join(path))
    print("Total Cost :", cost)
else:
    print("No path found!")
```
